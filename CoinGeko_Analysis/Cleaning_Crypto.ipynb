```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("C:\\Users\\vaish\\OneDrive\\Documents\\Advanced_Multi-Source_Data_Pipeline\\CoinGeko_Analysis\\Scraped_crypto(1).csv")
df
```

```python
# standardising the column name

df.columns=df.columns.str.strip().str.title()
```

```python
# ID column

# column values into title case
df["Id"]=df["Id"].str.strip().str.title()
df["Id"].isnull().sum()
df["Id"].duplicated().sum()
df["Id"].unique()

```

```python
# symbol column

df["Symbol"]=df["Symbol"].str.strip().str.title().astype("string")
df["Symbol"].isnull().sum()
df["Symbol"].duplicated().sum()
```

```python
# name column
df["Name"]=df["Name"].str.strip().str.title()
df["Name"].isnull().sum()
df["Name"].duplicated().sum()
```

```python
# Image column
df["Image"]=df["Image"].str.split()
df["Image"].isnull().sum()
df["Image"].duplicated().sum()
```

```python
# current price column
df["Current_Price"].isnull().sum()
df["Current_Price"].duplicated().sum()
df["Current_Price"].value_counts()

print("zero values=",(df["Current_Price"]==0).sum())

print("Negative values: ",(df["Current_Price"]< 0).sum())
```

```python
# market cap columnn\

df["Market_Cap"].isnull().sum()
df["Market_Cap"].duplicated().sum()
df["Market_Cap"].value_counts()
df["Market_Cap"].dtype

print("Zero value: ",(df["Market_Cap"]==0).sum())
print("negative value: ",(df["Market_Cap"]<0).sum())
```

```python
# Market_Cap_Rank column
print(df["Market_Cap_Rank"].isna().sum())

df["Market_Cap_Rank"].duplicated().sum()

print("Zero value: ",(df["Market_Cap_Rank"]==0).sum())

print("-ve value: ",(df["Market_Cap_Rank"]<0).sum())
```

```python
# Fully_Diluted_Valuation column

df["Fully_Diluted_Valuation"].isnull().sum()

df["Fully_Diluted_Valuation"].duplicated().sum()

print("Zero Value: ",(df["Fully_Diluted_Valuation"]==0).sum())
# convert 0 into the nan values
df["Fully_Diluted_Valuation"]=df["Fully_Diluted_Valuation"].replace(0,np.nan)
df["Fully_Diluted_Valuation"]

print("negative Value: ",(df["Fully_Diluted_Valuation"]<0).sum())

# check the missing values
print("Missing values: ",df["Fully_Diluted_Valuation"].isna().sum())
```

```python
# Total_Volume column
df["Total_Volume"].isnull().sum()

df["Total_Volume"].duplicated().sum()

print("Zero Value: ",(df["Total_Volume"]==0).sum())
# check their percentage
print("zero_per:",(df['Total_Volume'] == 0).mean()*100)
print("-"*50)

# replace 0 eith nan
df["Total_Volume"]=df["Total_Volume"].replace(0,np.nan)
print(df["Total_Volume"])

df["Total_Volume"].dtype
```
