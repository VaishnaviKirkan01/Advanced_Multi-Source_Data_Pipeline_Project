2. Bivariate Analysis

• Correlation heatmap (numerical features)

```python
numeric_cols=df.select_dtypes(include="number")
corr=numeric_cols.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Price vs Market Cap ralationship

sns.scatterplot(
    data=df,
    x="Current_Price",
    y="Market_Cap",
    palette="deep"
)
plt.title("Price vs Market Cap")
plt.xlabel("Current_Price")
plt.ylabel("Market_Cap")

plt.plot
```

```python
# price vs volume relationship

sns.scatterplot(
    data=df,
    x="Current_Price",
    y="Total_Volume",
    color="red"
)
plt.title("Price vs Total_Volume")
plt.xlabel("Current_Price")
plt.ylabel("Total_Volume")

plt.plot
```

```python
# adding the trend line using the regplot

sns.regplot(
    data=df,
    x='Market_Cap',
    y='Current_Price',
    color="green"
)

plt.title("Market Cap vs Current Price")
plt.xlabel("Market Cap")
plt.ylabel("Current Price")
plt.show()
```

```python
sns.regplot(
    data=df,
    x='Market_Cap',
    y='Total_Volume',
    color="cyan"
)

plt.title("Market Cap vs Total_Volume")
plt.xlabel("Market Cap")
plt.ylabel("Total_Volume")
plt.show()
```

3. Multivariate Analysis

Market Cap categories vs Price Change %

```python
df['Market_Cap_Category']=pd.qcut(
    df["Market_Cap"],q=3,labels=["Small","Medium","Large"]
)
```

```python
sns.boxplot(
    data=df,
    x='Market_Cap_Category',
    y="Price_Change_Percentage_24H",
    palette="deep",

)
plt.title("Market Cap Category vs Price Change %")
plt.xlabel("Market Cap Category")
plt.ylabel("Price Change %")
plt.plot
```

```python
columns=[
    "Current_Price",
    "Market_Cap",
    "Total_Volume",
    "Price_Change_Percentage_24H",

]

sns.pairplot(df[columns])
plt.show()
```

```python
df.groupby('Market_Cap_Category').agg({
    "Current_Price":["mean","median"],
    "Total_Volume":["mean","median"],
    "Price_Change_Percentage_24H":["mean","median"]
})
```

4. Advanced EDA

```python
#  outliers analysis
columns=[
    "Current_Price",
    "Market_Cap",
    "Total_Volume",
    "Price_Change_Percentage_24H"
]

for col in columns:
    sns.boxplot(
        data=df,
        x=col,
        palette="deep",

    )
    plt.title("Boxplot of the columns")
    plt.xlabel("Columns")
    plt.show()
```
