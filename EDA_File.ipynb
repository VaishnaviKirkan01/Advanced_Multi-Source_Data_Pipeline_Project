```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("C:\\Users\\vaish\\OneDrive\\Documents\\Advanced_Multi-Source_Data_Pipeline\\Cleaned_Books_excel.csv")
```

```python
#  Which categories are most expensive?  

sns.barplot(
    data=df,
    x=df['category'],
    y=df['price'],
    hue=df['category'],
    palette="pastel",

)
plt.title("Category Vs Price of Books")
plt.xlabel("Category")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```

```python
#  Which categories have the highest ratings?  

sns.barplot(
    data=df,
    x=df['category'],
    y=df['rating'],
    hue=df['category'],
    palette="pastel",

)
plt.title("Category Vs rating of Books")
plt.xlabel("Categories")
plt.ylabel("rating")
plt.xticks(rotation=45)
plt.show()
```

```python
#  Is there a relationship between price and rating?  
sns.boxenplot(
    data=df,
    x=df['rating'],
    y=df['price'],
    palette="deep",
    hue="rating"

)
plt.title("Rating VS Price")
plt.xlabel("Rating")
plt.ylabel("price")
plt.xticks(rotation=45)
plt.show()
```

```python
# • Which categories have the most books?  
sns.countplot(
    data=df,
    x=df["category"],
    palette='deep',
    hue='category'
)
plt.title("Book count graph")
plt.xlabel("Category")
plt.ylabel('count')
plt.xticks(rotation=45)
plt.show()
```

```python
# • What percentage of books are available?  


availability = (
    df["avalibility"].eq("In stock")
    .groupby(df["category"])
    .mean() * 100
)

sns.barplot(
    x=availability.index,
    y=availability.values,
    hue=availability.index,
    palette="pastel",
    legend=False
)

plt.title("Percentage of Books Available")
plt.xlabel("Category")
plt.ylabel("Availability (%)")
plt.xticks(rotation=45)
plt.show()
```
